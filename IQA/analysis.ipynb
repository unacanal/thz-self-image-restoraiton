```python
def get_max(metric='Sharpness'):
    print(f'MAX {metric}: ', end='')
    print(data.index[data[metric].argmax()], data[metric].max())

def get_min(metric='BRISQUE'):
    print(f'MIN {metric}: ', end='')
    print(data.index[data[metric].argmin()], data[metric].min())
```

```python
get_max('Sharpness')
get_max('Contrast')
get_min('BRISQUE')
```
