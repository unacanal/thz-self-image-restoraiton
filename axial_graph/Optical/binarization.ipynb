```python
import matplotlib.pyplot as plt
import cv2
import numpy as np
```

```python
from PIL import Image

temp = Image.open('segment anything.png')
len(temp.split())

# np.array(temp).shape#[1367:1700,500:1080]
# temp = Image.fromarray(255 - np.rot90(np.array(temp)[1367:1700,500:1080], 1))
image = np.rot90(np.array(temp), 1)
image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
thresh = 90
maxValue = 255 
th, dst = cv2.threshold(image, thresh, maxValue, cv2.THRESH_BINARY)
temp = Image.fromarray(dst)
temp.show()
```
